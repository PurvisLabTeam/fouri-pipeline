## About this notebook:

This notebook measures every cell in every round. Cells are outlined once, on the anchor round's segmentation from 01 (the first round, unless `experiment.json` sets `anchor_round`). For every other round, the outlines are moved to where the cells sit in that round, using the round's transform from 02 (or 03), and each channel is measured on 00's images. Several wells run at once, about 10–20 seconds per well.

Input:
- `segmented/{well}/`: the anchor round's cell outlines from 01
- `output_df/round_table_{well}.csv` from 00 and `output_df/tmat_{well}.npy` from 02 (or 03)
- `output_images/{well}/`: every channel's TIFF from 00. Experiments run through an older version of 00 need 00 run again first.

Settings:
- `ring_width`, `ring_gap`: the cytoplasmic ring around each nucleus reaches `ring_width` pixels out and starts `ring_gap + 1` pixels from the nucleus edge (6 and 1 by default). Pixels within `ring_width` of two nuclei belong to neither ring. Use the same values for experiments you want to compare.

Output: `cell_data/cell_data_{well}_df.csv`, one row per cell:
- shape, from the anchor round: `label`, `nuc_area`, `centroid-0`, `centroid-1`, `orientation`, `axis_major_length`, `axis_minor_length`, `bbox-0` to `bbox-3`, `ring_area`
- for every round and channel, e.g. `R0_pRB`: `_nuc_mean`, `_ring_mean`, `_total_nuc_protein` (mean × nucleus area) and `_nuc_cyto_ratio` (nucleus mean ÷ ring mean)
- `well`

Left out: cells touching the image edge in the anchor round, cells with no ring, and cells that fall outside a round's image. Masks from 05 are not used in this version.

Results differ slightly from versions before November 2026, which moved the images instead of the outlines, so don't combine older and newer `cell_data` in 07.
