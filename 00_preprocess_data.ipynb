## About this notebook

This notebook turns the microscope's ND2 files into one TIFF per channel, ready for segmentation (01) and alignment (02). It runs every well in the experiment, several wells at once. All wells must have been imaged with the same rounds and antibodies, since one `info_exp.csv` describes them all; process wells imaged differently as separate experiments.

Input:
- the stitched ND2 files in `data/`, one subfolder per round (see the README for the folder layout)
- `info_exp.csv`: one row per round and one column per optical configuration; each cell names the signal (antibody or stain) imaged in that round, or is left empty

Setting:
- `align_signal`: the signal the rounds are aligned on, written exactly as in `info_exp.csv` (usually `DNA`). Every round must have exactly one channel with this signal.

Output:
- `output_images/{well}/`: every channel of every round as a compressed TIFF, `Round_{round}_well{well}_{signal}.tif`, all cut to the size of the smallest image. Each file keeps the pixel size, wavelengths and exposure time recorded by the microscope. About 0.3 to 0.9 GB per well, depending on how much of the image is cells.
- `output_df/round_table_{well}.csv`: one row per round and channel, listing the file it came from, its signal and its acquisition settings.

Experiments run through an older version of this notebook need it run again before 01.
